# KLA Hackathon - Train on A100 GPU

## How to get an A100 GPU in Google Colab:
1. In the top menu, click **Runtime** > **Change runtime type**.
2. Under **Hardware accelerator**, select **GPU**.
3. Under **GPU class**, select **A100** (Note: A100 requires Colab Pro or Colab Pro+). If you don't have Colab Pro, you can select T4 or V100, which will still work perfectly but will be slightly slower.
4. Click **Save**.

## Other places to get A100 GPUs:
If you want to train outside of Colab, you can rent A100 instances cheaply on:
- **RunPod.io** (~$0.80/hour for A100)
- **Lambda Labs** (~$1.10/hour for A100)
- **AWS (p4d instances)** or **Google Cloud (a2 instances)** (More expensive, enterprise focus)
